## 1. Marco conceptual

### 1.1 ¿Qué es el análisis predictivo?

La Guía Analítica Empresarial distingue tres tipos de análisis según la pregunta que responden (Universidad de Cundinamarca, 2024). En la práctica anterior se trabajó el análisis descriptivo; en esta se avanza al **predictivo**:

| Tipo de análisis | Pregunta clave | Propósito | Ejemplo |
|---|---|---|---|
| Descriptivo | ¿Qué ha sucedido? | Comprender el pasado | Informes de ventas |
| **Predictivo** | ¿Qué podría suceder? | Anticipar escenarios futuros | Predicción de la demanda |


Google Cloud (s.f.) define el análisis predictivo como el proceso de usar **datos históricos, modelos estadísticos y aprendizaje automático** para predecir resultados futuros. Su pregunta guía es *«¿Qué podría ocurrir a continuación?»*. A diferencia del descriptivo, no se queda en explicar el pasado: busca patrones en los datos para estimar lo que es probable que ocurra y así apoyar decisiones con anticipación.

### 1.2 Diferencias entre el análisis descriptivo y el predictivo

En la práctica anterior (análisis descriptivo de *TecnoAndes*) se respondió **qué sucedió**. En esta se usa el histórico para estimar **qué podría suceder**. La tabla resume en qué cambia el enfoque:

| Aspecto | Análisis descriptivo | Análisis predictivo |
|---|---|---|
| Pregunta | ¿Qué ha sucedido? | ¿Qué podría suceder? |
| Horizonte | Pasado y presente | Futuro y escenarios hipotéticos |
| Resultado | Resúmenes, indicadores y gráficos | Estimaciones, probabilidades y clasificaciones |
| Herramientas | Agregaciones, estadísticas, visualizaciones | Modelos estadísticos y de aprendizaje automático |
| Validación | Conciliación y coherencia de los datos | Medición del error con datos que el modelo no conoce |
| Pregunta típica en este caso | ¿Cuánto se vendió en 2024? | ¿Cuánto se venderá en 2025 con un presupuesto dado? |

Ambos análisis son complementarios: el descriptivo (secciones 5 a 7 de este notebook) es el punto de partida obligado del predictivo, porque permite entender los datos y detectar errores antes de modelar (Universidad de Cundinamarca, 2024).

### 1.4 Técnicas utilizadas

Google Cloud (s.f.) destaca tres familias de técnicas: el **análisis de regresión**, que estima cómo un factor (por ejemplo, el precio o la inversión) afecta a otro (por ejemplo, las ventas); los **árboles de decisión**, que clasifican los datos en categorías siguiendo una estructura de ramas con reglas simples; y las **redes neuronales**, útiles para relaciones muy complejas. ## 6. Limpieza y preparación de los datos

La limpieza corresponde al paso de **preprocesamiento** del proceso predictivo (Google Cloud, s.f.). Además de corregir el error, se crean las variables que necesitan los modelos:

| Paso | Problema / necesidad | Regla aplicada |
|---|---|---|
| 1 | Años 2068 – 2079 imposibles | Reemplazar por 2024 los años mayores a 2023 (continúan la secuencia 2020 – 2023) |
| 2 | `Mes` en texto no permite ordenar | Crear `Mes_num` (1 a 12) y una columna `Fecha` |
| 3 | La regresión logística necesita una variable binaria | Crear `Meta_cumplida` = 1 si las ventas superan el umbral (sección 10) |
| 4 | El árbol de decisión necesita categorías | Crear `Categoria_ventas` (bajas, medias, altas) por terciles (sección 11) |

En esta sección se aplican los pasos 1 y 2; los pasos 3 y 4 se explican junto a cada modelo.

```python
ventas = ventas_raw.copy()

# Paso 1: corrección de los años imposibles
años_erroneos = ventas.loc[ventas["Año"] > 2023, "Año"].tolist()
ventas.loc[ventas["Año"] > 2023, "Año"] = 2024
print(f"Años corregidos ({len(años_erroneos)}): {años_erroneos} -> 2024")

# Paso 2: número de mes y fecha
MESES = {"Enero": 1, "Febrero": 2, "Marzo": 3, "Abril": 4, "Mayo": 5, "Junio": 6,
         "Julio": 7, "Agosto": 8, "Septiembre": 9, "Octubre": 10, "Noviembre": 11, "Diciembre": 12}
ventas["Mes_num"] = ventas["Mes"].map(MESES)
ventas["Fecha"] = pd.to_datetime(dict(year=ventas["Año"], month=ventas["Mes_num"], day=1))

# --- Validaciones: si alguna regla no se cumple, la celda se detiene con un error ---
assert ventas["Año"].between(2020, 2024).all(), "Hay años fuera de rango"
assert (ventas.groupby("Año").size() == 12).all(), "Algún año no tiene 12 meses"
assert ventas["Mes_num"].notna().all(), "Hay meses sin traducir"
assert ventas["Fecha"].is_unique, "Hay periodos repetidos"
assert ventas["Promociones_aplicadas"].isin([0, 1]).all()
assert ventas["Opinion_clientes"].between(1, 5).all()
print("Validaciones superadas: 5 años completos, 60 periodos únicos")
ventas.groupby("Año").size().rename("meses").to_frame().T
```

### 7.1 Medidas de tendencia central y de dispersión

La guía pide calcular medidas de **tendencia central** (media, mediana, moda) y de **dispersión** (desviación estándar, coeficiente de variación, rango intercuartílico). La función `resumen_estadistico()` las reúne junto con la forma de la distribución (asimetría y curtosis).

```python
def resumen_estadistico(serie):
    """Medidas de tendencia central, dispersión y forma de una variable numérica."""
    q1, q3 = serie.quantile([0.25, 0.75])
    return pd.Series({
        "n": serie.count(),
        "media": serie.mean(),
        "mediana": serie.median(),
        "moda": serie.mode().iloc[0],
        "desv_estandar": serie.std(),
        "CV (%)": serie.std() / serie.mean() * 100,
        "minimo": serie.min(),
        "Q1": q1,
        "Q3": q3,
        "RIC": q3 - q1,
        "maximo": serie.max(),
        "asimetria": stats.skew(serie),
        "curtosis": stats.kurtosis(serie),
    })

VARIABLES = ["Ventas_mensuales", "Presupuesto_marketing", "Nuevos_clientes",
             "Promociones_aplicadas", "Opinion_clientes"]
tabla_resumen = ventas[VARIABLES].apply(resumen_estadistico).T
tabla_resumen
```

```python
# Ventas promedio por año: ¿hay tendencia en el tiempo?
ventas.groupby("Año").agg(
    ventas_promedio=("Ventas_mensuales", "mean"),
    ventas_totales=("Ventas_mensuales", "sum"),
    presupuesto_promedio=("Presupuesto_marketing", "mean"),
    meses_con_promocion=("Promociones_aplicadas", "sum"),
)
```

```python
# Frecuencias de las variables cualitativas
for col, etiqueta in [("Promociones_aplicadas", "Promociones (0 = no, 1 = sí)"),
                      ("Opinion_clientes", "Opinión de clientes (1 a 5)")]:
    freq = ventas[col].value_counts().sort_index().to_frame("meses")
    freq["%"] = freq["meses"] / len(ventas) * 100
    print(f"\n{etiqueta}\n{freq.to_string()}")
```

**Interpretación.**

* **Ventas mensuales:** la media (≈ 7,467) y la mediana (≈ 7,469) son prácticamente iguales y la asimetría es cercana a cero, lo que indica una distribución **simétrica**. El coeficiente de variación es moderado (≈ 19 %): las ventas cambian de un mes a otro, pero sin valores extremos.
* **Presupuesto de marketing:** tiene un comportamiento muy parecido al de las ventas (media ≈ 4,845 y CV similar), primera señal de que ambas variables se mueven juntas.
* **Nuevos clientes:** oscilan entre ≈ 50 y ≈ 200 por mes (media ≈ 143), con mayor dispersión relativa (CV ≈ 30 %) y una ligera asimetría negativa.
* **Nota sobre la moda:** en las variables continuas cada valor aparece una sola vez, así que la moda solo corresponde al valor mínimo y no es informativa; en ellas la media y la mediana son las medidas de tendencia central adecuadas. La moda sí es útil en las variables discretas (promociones = 1, opinión = 2).
* **Promociones:** se aplicaron en el 55 % de los meses, por lo que las dos situaciones (con y sin promoción) están bien representadas.
* **Opinión de clientes:** el promedio es bajo (≈ 2.8 sobre 5), un punto de atención para el negocio que es independiente de la predicción.
* Por año, las ventas promedio suben y bajan (de ≈ 8,188 en 2020 a ≈ 6,781 en 2021 y ≈ 7,811 en 2024) sin una tendencia sostenida, y cada cambio acompaña al presupuesto promedio de ese año. Esto respalda no usar el año como predictor.

### 7.2 Relación entre las variables

La correlación mide la fuerza y la dirección de la relación entre dos variables (de −1 a 1). Se calculan dos coeficientes: **Pearson**, para relaciones lineales, y **Spearman**, basado en rangos y adecuado para variables ordinales como la opinión.

```python
correlaciones = pd.DataFrame({
    "Pearson": ventas[VARIABLES].corr(method="pearson")["Ventas_mensuales"],
    "Spearman": ventas[VARIABLES].corr(method="spearman")["Ventas_mensuales"],
}).drop("Ventas_mensuales").sort_values("Pearson", ascending=False)

# Valor p de Pearson: ¿la correlación es estadísticamente distinta de cero?
correlaciones["p_valor"] = [stats.pearsonr(ventas[v], ventas["Ventas_mensuales"])[1]
                            for v in correlaciones.index]
correlaciones["¿significativa? (α=0.05)"] = np.where(correlaciones["p_valor"] < 0.05, "Sí", "No")
correlaciones.style.format({"Pearson": "{:.3f}", "Spearman": "{:.3f}", "p_valor": "{:.4f}"})
```

```python
fig, ax = plt.subplots(figsize=(7, 5.5))
matriz = ventas[VARIABLES].corr()
mascara = np.triu(np.ones_like(matriz, dtype=bool), k=1)   # solo el triángulo inferior
sns.heatmap(matriz, mask=mascara, annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1,
            linewidths=0.5, cbar_kws={"shrink": 0.8}, ax=ax)
ax.set_title("Matriz de correlación de Pearson")
plt.tight_layout()
plt.show()
```

**Interpretación.**

* El **presupuesto de marketing** tiene una correlación **muy fuerte y positiva** con las ventas (r ≈ 0.98, significativa). Es, con diferencia, la variable más relacionada con el resultado.
* **Nuevos clientes, promociones y opinión** muestran correlaciones débiles con las ventas y, en general, no significativas. En este dataset su aporte individual a la explicación de las ventas es reducido.

## 8. Selección de variables relevantes

Antes de entrenar los modelos se decide **qué variables entran en cada uno**. Seleccionar bien las variables evita incluir ruido, hace los modelos más fáciles de explicar y reduce el riesgo de sobreajuste. Se aplican tres criterios:

| Criterio | Prueba | Qué indica |
|---|---|---|
| Relación con las ventas | Correlación de Pearson y Spearman (variables numéricas) | Si la variable se mueve junto con las ventas |
| Diferencia entre grupos | Prueba *t* de Welch (promociones) y ANOVA / Kruskal-Wallis (opinión) | Si las ventas promedio cambian según el grupo |
| Redundancia entre predictores | Factor de inflación de la varianza (VIF) | Si dos predictores aportan la misma información (VIF > 5 es problemático) |

```python
PREDICTORES = ["Presupuesto_marketing", "Nuevos_clientes", "Promociones_aplicadas", "Opinion_clientes"]
y_ventas = ventas["Ventas_mensuales"]

# Criterio 1: correlación con las ventas
r_pearson = {v: stats.pearsonr(ventas[v], y_ventas) for v in PREDICTORES}

# Criterio 2: diferencia de ventas entre grupos
con_promo = y_ventas[ventas["Promociones_aplicadas"] == 1]
sin_promo = y_ventas[ventas["Promociones_aplicadas"] == 0]
p_promo = stats.ttest_ind(con_promo, sin_promo, equal_var=False).pvalue
grupos_opinion = [g["Ventas_mensuales"] for _, g in ventas.groupby("Opinion_clientes")]
p_opinion_anova = stats.f_oneway(*grupos_opinion).pvalue
p_opinion_kw = stats.kruskal(*grupos_opinion).pvalue

# Criterio 3: redundancia entre predictores (VIF)
X_vif = sm.add_constant(ventas[PREDICTORES])
vif = {v: variance_inflation_factor(X_vif.values, i + 1) for i, v in enumerate(PREDICTORES)}

seleccion = pd.DataFrame({
    "r_Pearson": {v: r_pearson[v][0] for v in PREDICTORES},
    "p_Pearson": {v: r_pearson[v][1] for v in PREDICTORES},
    "p_prueba_grupos": {"Presupuesto_marketing": np.nan, "Nuevos_clientes": np.nan,
                        "Promociones_aplicadas": p_promo, "Opinion_clientes": p_opinion_kw},
    "VIF": vif,
})
seleccion["¿relación significativa?"] = np.where(
    seleccion[["p_Pearson", "p_prueba_grupos"]].min(axis=1) < 0.05, "Sí", "No")

print(f"Promociones – diferencia de ventas promedio (con − sin): {con_promo.mean() - sin_promo.mean():,.0f} "
      f"(prueba t de Welch, p = {p_promo:.3f})")
print(f"Opinión – ANOVA p = {p_opinion_anova:.3f} | Kruskal-Wallis p = {p_opinion_kw:.3f}")
seleccion.style.format({"r_Pearson": "{:.3f}", "p_Pearson": "{:.4f}", "p_prueba_grupos": "{:.4f}", "VIF": "{:.2f}"}, na_rep="—")
```

**Interpretación y decisión.**

* El **presupuesto de marketing** es la única variable con relación significativa con las ventas (r ≈ 0.98, p < 0.001).
* **Nuevos clientes** no se correlaciona con las ventas (r ≈ −0.01). Los meses **con promoción** venden en promedio solo ≈ 150 unidades más que los meses sin promoción, una diferencia que no es significativa (p ≈ 0.68). Las ventas tampoco cambian según la **opinión** de los clientes (p > 0.4).
* Todos los VIF son cercanos a 1: los predictores no se repiten información entre sí, así que pueden usarse juntos sin problemas de multicolinealidad.

Con base en esto y en lo que pide la guía del taller, la selección queda así:

| Modelo | Variables seleccionadas | Justificación |
|---|---|---|
| Regresión lineal (P1) | Presupuesto de marketing | La pregunta P1 es sobre el efecto del presupuesto, y es la única variable significativa |
| Regresión logística (P2) | Las 4 variables | La pregunta P2 busca identificar qué factores influyen; se incluyen todas para **comprobar** su aporte dentro del modelo |
| Árbol de decisión (P3) | Las 4 variables | El árbol hace su propia selección: usa solo las variables que mejoran la clasificación y descarta las demás |
| Variables temporales | Ninguna | No se gestionan desde el negocio y no muestran un patrón estable (sección 6.1) |

Que una variable no sea significativa también es un hallazgo: la empresa invierte en **promociones** en el 55 % de los meses sin que se refleje en mayores ventas.

## 9. Modelo 1 – Regresión lineal simple (P1)

### 9.1 Planteamiento

**Pregunta:** ¿cómo afecta el presupuesto de marketing a las ventas mensuales?

La regresión lineal estima una recta que relaciona una variable independiente con una dependiente (Google Cloud, s.f.):

$$\text{Ventas}_{mensuales} = \beta_0 + \beta_1 \cdot \text{Presupuesto}_{marketing} + \varepsilon$$

* $\beta_0$ (intercepto): ventas esperadas cuando el presupuesto es cero (valor teórico).
* $\beta_1$ (pendiente): cambio esperado en las ventas por cada unidad adicional invertida en marketing.

**Validación.** Los datos se dividen en **entrenamiento (70 %)** para ajustar el modelo y **prueba (30 %)** para medir su precisión con meses que el modelo no conoce. Las métricas que pide la guía son:

| Métrica | Qué mide | Mejor valor |
|---|---|---|
| **R²** | Proporción de la variación de las ventas que explica el modelo | Cercano a 1 |
| **MSE** | Error cuadrático medio (en unidades al cuadrado) | Cercano a 0 |
| **RMSE** | Raíz del MSE: error típico en las mismas unidades de las ventas | Cercano a 0 |
| MAE | Error absoluto medio | Cercano a 0 |

```python
# Variables del modelo
X_lin = ventas[["Presupuesto_marketing"]]
y_lin = ventas["Ventas_mensuales"]

# División entrenamiento / prueba
X_tr_lin, X_te_lin, y_tr_lin, y_te_lin = train_test_split(
    X_lin, y_lin, test_size=0.30, random_state=SEMILLA)

# Entrenamiento
modelo_lineal = LinearRegression().fit(X_tr_lin, y_tr_lin)
b0, b1 = modelo_lineal.intercept_, modelo_lineal.coef_[0]

print(f"Registros de entrenamiento: {len(X_tr_lin)} | de prueba: {len(X_te_lin)}")
print(f"\nEcuación estimada:  Ventas = {b0:,.2f} + {b1:.4f} × Presupuesto_marketing")
```

```python
# Significancia del coeficiente con statsmodels (mismos datos de entrenamiento)
ols = sm.OLS(y_tr_lin, sm.add_constant(X_tr_lin)).fit()
pd.DataFrame({
    "coeficiente": ols.params,
    "error_estandar": ols.bse,
    "t": ols.tvalues,
    "p_valor": ols.pvalues,
    "IC 95% inf": ols.conf_int()[0],
    "IC 95% sup": ols.conf_int()[1],
}).style.format("{:,.4f}")
```

```python
def metricas_regresion(y_real, y_pred):
    """R², MSE, RMSE y MAE de un modelo de regresión."""
    mse = mean_squared_error(y_real, y_pred)
    return {"R²": r2_score(y_real, y_pred), "MSE": mse, "RMSE": np.sqrt(mse),
            "MAE": mean_absolute_error(y_real, y_pred)}

pred_tr_lin = modelo_lineal.predict(X_tr_lin)
pred_te_lin = modelo_lineal.predict(X_te_lin)

metricas_lin = pd.DataFrame({
    "Entrenamiento": metricas_regresion(y_tr_lin, pred_tr_lin),
    "Prueba": metricas_regresion(y_te_lin, pred_te_lin),
})

# Validación cruzada (5 particiones): estabilidad del R² con diferentes divisiones
cv_r2_lin = cross_val_score(LinearRegression(), X_lin, y_lin, scoring="r2",
                            cv=KFold(5, shuffle=True, random_state=SEMILLA))
print(f"R² en validación cruzada (5 particiones): {cv_r2_lin.mean():.3f} ± {cv_r2_lin.std():.3f}")
print(f"Error relativo típico (RMSE / ventas promedio): "
      f"{metricas_lin.loc['RMSE', 'Prueba'] / y_lin.mean() * 100:.1f} %")
metricas_lin
```

```python
residuos = y_te_lin - pred_te_lin
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))

# Recta de regresión sobre los datos
ax1.scatter(X_tr_lin, y_tr_lin, color=AZUL, alpha=0.6, s=30, label="Entrenamiento")
ax1.scatter(X_te_lin, y_te_lin, color=NARANJA, s=45, label="Prueba")
xs = np.linspace(X_lin.min().iloc[0], X_lin.max().iloc[0], 100).reshape(-1, 1)
ax1.plot(xs, modelo_lineal.predict(pd.DataFrame(xs, columns=["Presupuesto_marketing"])),
         color="black", lw=1.5, label="Recta estimada")
ax1.set_title(f"Ventas vs. presupuesto de marketing (R² prueba = {metricas_lin.loc['R²', 'Prueba']:.3f})")
ax1.set_xlabel("Presupuesto de marketing"); ax1.set_ylabel("Ventas mensuales")
formato_miles(ax1.xaxis); formato_miles(ax1.yaxis)
ax1.legend(frameon=False)

# Residuos: diferencia entre lo real y lo predicho
ax2.scatter(pred_te_lin, residuos, color=NARANJA, s=45)
ax2.axhline(0, color="black", lw=1)
ax2.set_title("Residuos en los datos de prueba")
ax2.set_xlabel("Ventas predichas"); ax2.set_ylabel("Real − predicho")
formato_miles(ax2.xaxis); formato_miles(ax2.yaxis)
plt.tight_layout()
plt.show()
```

Por ejemplo, subir el presupuesto en 1,000 se asocia con unas 1,580 unidades más de ventas. El coeficiente es altamente significativo (p < 0.001).

## 10. Modelo 2 – Regresión logística (P2)

### 10.1 Planteamiento

**Pregunta:** ¿qué factores determinan si se alcanza la meta de ventas?

La regresión logística estima la **probabilidad** de que ocurra un evento binario (sí / no). Aquí el evento es *cumplir la meta de ventas del mes*.

**Definición de la variable objetivo.** La guía pide una variable binaria que valga 1 cuando las ventas superan un umbral. Se elige como umbral la **mediana histórica de las ventas** porque:

1. Representa un mes "típico": superar la mediana significa vender por encima de lo habitual, una meta exigente pero alcanzable.
2. Produce **clases balanceadas** (50 % / 50 %), lo que evita que el modelo aprenda a predecir siempre la clase mayoritaria.

**Variables independientes:** presupuesto de marketing, nuevos clientes, promociones y opinión de clientes. Como están en escalas muy distintas, se **estandarizan** (media 0 y desviación 1) para que los coeficientes sean comparables entre sí.

```python
# Paso 3 de la limpieza: variable objetivo binaria
UMBRAL_META = ventas["Ventas_mensuales"].median()
ventas["Meta_cumplida"] = (ventas["Ventas_mensuales"] > UMBRAL_META).astype(int)

print(f"Umbral de la meta (mediana de ventas): {UMBRAL_META:,.2f}")
balance = ventas["Meta_cumplida"].value_counts().rename({0: "No cumplió (0)", 1: "Cumplió (1)"}).to_frame("meses")
balance["%"] = balance["meses"] / len(ventas) * 100
balance
```

```python
# Variables seleccionadas en la sección 8
X_log = ventas[PREDICTORES]
y_log = ventas["Meta_cumplida"]

# División estratificada: conserva la proporción 50/50 en entrenamiento y prueba
X_tr_log, X_te_log, y_tr_log, y_te_log = train_test_split(
    X_log, y_log, test_size=0.30, random_state=SEMILLA, stratify=y_log)

# Pipeline: estandarización + regresión logística
modelo_logistico = make_pipeline(StandardScaler(), LogisticRegression(random_state=SEMILLA))
modelo_logistico.fit(X_tr_log, y_tr_log)

# Coeficientes estandarizados y razón de momios (odds ratio)
coef_log = pd.DataFrame({
    "coeficiente_estandarizado": modelo_logistico[-1].coef_[0],
    "odds_ratio (por 1 desv. est.)": np.exp(modelo_logistico[-1].coef_[0]),
}, index=PREDICTORES).sort_values("coeficiente_estandarizado", key=abs, ascending=False)
coef_log
```

```python
# ¿Cuánto aporta cada factor? Comparación de modelos con distintos grupos de variables.
# Nota: no se usan valores p clásicos porque el presupuesto separa casi perfectamente
# las dos clases ("separación cuasi completa"); en ese caso los errores estándar del
# modelo sin regularizar se disparan y los valores p dejan de ser confiables.
# scikit-learn aplica por defecto regularización L2, que mantiene los coeficientes estables.
grupos = {
    "Las 4 variables": PREDICTORES,
    "Solo presupuesto": ["Presupuesto_marketing"],
    "Sin presupuesto (3 variables)": [v for v in PREDICTORES if v != "Presupuesto_marketing"],
}
cv5 = StratifiedKFold(5, shuffle=True, random_state=SEMILLA)
aporte = pd.DataFrame({
    nombre: {
        "accuracy_validacion_cruzada": cross_val_score(
            make_pipeline(StandardScaler(), LogisticRegression()), ventas[vars_], y_log, cv=cv5).mean(),
        "AUC_validacion_cruzada": cross_val_score(
            make_pipeline(StandardScaler(), LogisticRegression()), ventas[vars_], y_log, cv=cv5,
            scoring="roc_auc").mean(),
    } for nombre, vars_ in grupos.items()
}).T
aporte
```

**Lectura de los coeficientes.** Un coeficiente positivo aumenta la probabilidad de cumplir la meta y uno negativo la reduce. El *odds ratio* indica cuántas veces se multiplican las posibilidades (*odds*) de cumplir la meta cuando la variable aumenta una desviación estándar. Un valor cercano a 1 significa que la variable casi no influye.

```python
def metricas_clasificacion(y_real, y_pred, promedio="binary"):
    """Accuracy, precisión, recall y F1 de un modelo de clasificación."""
    return {"Accuracy": accuracy_score(y_real, y_pred),
            "Precisión": precision_score(y_real, y_pred, average=promedio),
            "Recall": recall_score(y_real, y_pred, average=promedio),
            "F1": f1_score(y_real, y_pred, average=promedio)}

pred_te_log = modelo_logistico.predict(X_te_log)
prob_te_log = modelo_logistico.predict_proba(X_te_log)[:, 1]
auc_log = roc_auc_score(y_te_log, prob_te_log)

metricas_log = pd.DataFrame({
    "Entrenamiento": metricas_clasificacion(y_tr_log, modelo_logistico.predict(X_tr_log)),
    "Prueba": metricas_clasificacion(y_te_log, pred_te_log),
})
metricas_log.loc["AUC"] = [roc_auc_score(y_tr_log, modelo_logistico.predict_proba(X_tr_log)[:, 1]), auc_log]

cv_acc_log = cross_val_score(modelo_logistico, X_log, y_log, scoring="accuracy",
                             cv=StratifiedKFold(5, shuffle=True, random_state=SEMILLA))
print(f"Accuracy en validación cruzada (5 particiones): {cv_acc_log.mean():.3f} ± {cv_acc_log.std():.3f}")
metricas_log
```

* **Nuevos clientes, promociones y opinión** tienen coeficientes pequeños (odds ratio cercanos a 1). La tabla de aporte lo confirma: el modelo **solo con el presupuesto** predice igual o mejor que con las cuatro variables, mientras que **sin el presupuesto** el acierto cae a niveles cercanos al azar (≈ 50 %). En este histórico, esas tres variables no cambian la probabilidad de cumplir la meta. El signo ligeramente negativo de las promociones no debe leerse como un efecto real, sino como ruido de una muestra pequeña.
* **Precisión:** en los datos de prueba el modelo acierta el **94 %** de los meses (17 de 18). La **precisión** es del 90 % (de los meses que predijo como "cumplidos", 9 de 10 lo fueron) y el **recall** del 100 % (detectó todos los meses que sí cumplieron). El **AUC ≈ 0.99** indica una capacidad casi perfecta para separar ambos grupos. La validación cruzada (≈ 0.90) confirma un desempeño alto y estable.
* La matriz de confusión muestra **un solo error**: un mes que no cumplió la meta fue clasificado como cumplido (falso positivo), un error de optimismo que conviene tener presente.
* **Lectura para el negocio:** el modelo permite calcular, antes de iniciar el mes, la **probabilidad de alcanzar la meta** según el presupuesto aprobado. Con promociones activas y valores típicos en las demás variables, la probabilidad supera el 50 % a partir de un presupuesto de **≈ 4,970**, un dato útil para fijar el presupuesto mínimo recomendado.

## 11. Modelo 3 – Árbol de decisión (P3)

### 11.1 Planteamiento

**Pregunta:** ¿se puede predecir si las ventas de un mes serán bajas, medias o altas a partir de varias variables?

Un árbol de decisión clasifica los datos mediante una secuencia de preguntas del tipo *«¿el presupuesto es mayor que X?»*, organizadas en forma de ramas (Google Cloud, s.f.). Su principal ventaja es que las reglas son **fáciles de explicar** a personas sin formación técnica.

**Definición de las categorías.** Las ventas se dividen en **terciles** (tres grupos con el mismo número de meses):

| Categoría | Regla |
|---|---|
| Bajas | Ventas ≤ percentil 33.3 |
| Medias | Entre el percentil 33.3 y el 66.7 |
| Altas | Ventas > percentil 66.7 |

Se usan terciles porque generan clases balanceadas (20 meses cada una) y porque reflejan el desempeño **relativo** de la empresa frente a su propio historial.

**Control del sobreajuste.** Un árbol sin límite puede memorizar los datos de entrenamiento. Con solo 60 registros se limita la **profundidad máxima a 3 niveles**; la tabla de comparación de profundidades justifica esta elección.

```python
# Paso 4 de la limpieza: variable objetivo de tres categorías (terciles)
ventas["Categoria_ventas"], cortes = pd.qcut(ventas["Ventas_mensuales"], q=3,
                                             labels=["Bajas", "Medias", "Altas"], retbins=True)
print("Límites de las categorías:", [f"{c:,.0f}" for c in cortes])
ventas.groupby("Categoria_ventas", observed=True)["Ventas_mensuales"].agg(["count", "min", "max", "mean"])
```

```python
X_arb = ventas[PREDICTORES]
y_arb = ventas["Categoria_ventas"]

X_tr_arb, X_te_arb, y_tr_arb, y_te_arb = train_test_split(
    X_arb, y_arb, test_size=0.30, random_state=SEMILLA, stratify=y_arb)

# Comparación de profundidades: accuracy en entrenamiento, prueba y validación cruzada
filas = []
for prof in [1, 2, 3, 4, 5, None]:
    arbol = DecisionTreeClassifier(max_depth=prof, random_state=SEMILLA).fit(X_tr_arb, y_tr_arb)
    cv = cross_val_score(DecisionTreeClassifier(max_depth=prof, random_state=SEMILLA), X_arb, y_arb,
                         cv=StratifiedKFold(5, shuffle=True, random_state=SEMILLA))
    filas.append({"profundidad": "sin límite" if prof is None else prof,
                  "accuracy_entrenamiento": arbol.score(X_tr_arb, y_tr_arb),
                  "accuracy_prueba": arbol.score(X_te_arb, y_te_arb),
                  "accuracy_validacion_cruzada": cv.mean()})
pd.DataFrame(filas).set_index("profundidad")
```

**Elección de la profundidad.** Con 1 nivel el árbol es demasiado simple (no logra separar tres clases). A partir de 4 niveles acierta casi todo en entrenamiento pero empeora en prueba y en validación cruzada: es la señal típica del **sobreajuste**. Las profundidades 2 y 3 tienen un desempeño similar en validación cruzada; se elige **3** porque logra la mejor exactitud en prueba con reglas todavía sencillas de leer.

```python
modelo_arbol = DecisionTreeClassifier(max_depth=3, random_state=SEMILLA).fit(X_tr_arb, y_tr_arb)
pred_te_arb = modelo_arbol.predict(X_te_arb)

metricas_arb = pd.DataFrame({
    "Entrenamiento": metricas_clasificacion(y_tr_arb, modelo_arbol.predict(X_tr_arb), promedio="macro"),
    "Prueba": metricas_clasificacion(y_te_arb, pred_te_arb, promedio="macro"),
})
cv_acc_arb = cross_val_score(DecisionTreeClassifier(max_depth=3, random_state=SEMILLA), X_arb, y_arb,
                             cv=StratifiedKFold(5, shuffle=True, random_state=SEMILLA))
print(f"Accuracy en validación cruzada (5 particiones): {cv_acc_arb.mean():.3f} ± {cv_acc_arb.std():.3f}")
print("(Precisión, recall y F1 son promedios macro de las tres categorías)")
metricas_arb
```

El color indica la clase dominante y su intensidad, qué tan "pura" es la hoja.

## 13. Proyección de escenarios para 2025 (P5)

El fin último del análisis predictivo es **anticipar lo que podría ocurrir** para decidir con tiempo (Google Cloud, s.f.). Esta sección usa los modelos para proyectar las ventas de 2025 bajo tres escenarios de inversión en marketing.

**Modelos finales.** Una vez validados con los datos de prueba, los modelos se **reentrenan con los 60 meses** para aprovechar toda la información disponible. Esta es la práctica habitual: la división entrenamiento / prueba sirve para medir la precisión; el modelo que se usa para proyectar aprende de todo el histórico.

**Escenarios.** Se toma como base el presupuesto mes a mes de 2024 y se ajusta en tres niveles. Las demás variables (nuevos clientes, promociones y opinión) se mantienen como en 2024:

| Escenario | Presupuesto mensual 2025 |
|---|---|
| Conservador | 10 % menos que en 2024 |
| Base | Igual que en 2024 |
| Expansión | 10 % más que en 2024 |

```python
# --- Modelos finales reentrenados con los 60 meses --------------------------------
ols_final = sm.OLS(ventas["Ventas_mensuales"], sm.add_constant(ventas[["Presupuesto_marketing"]])).fit()
logistico_final = make_pipeline(StandardScaler(), LogisticRegression(random_state=SEMILLA)).fit(
    ventas[PREDICTORES], ventas["Meta_cumplida"])
arbol_final = DecisionTreeClassifier(max_depth=3, random_state=SEMILLA).fit(
    ventas[PREDICTORES], ventas["Categoria_ventas"])
print(f"Ecuación final (60 meses): Ventas = {ols_final.params['const']:,.2f} + "
      f"{ols_final.params['Presupuesto_marketing']:.4f} × Presupuesto  |  R² = {ols_final.rsquared:.3f}")

# --- Escenarios 2025 ---------------------------------------------------------------
base_2024 = ventas[ventas["Año"] == 2024].reset_index(drop=True)
ESCENARIOS = {"Conservador": 0.90, "Base": 1.00, "Expansión": 1.10}

def proyectar(factor):
    """Proyecta los 12 meses de 2025 con el presupuesto 2024 multiplicado por 'factor'."""
    x = base_2024[PREDICTORES].copy()
    x["Presupuesto_marketing"] = x["Presupuesto_marketing"] * factor
    X_ols = sm.add_constant(x[["Presupuesto_marketing"]], has_constant="add")
    pred = ols_final.get_prediction(X_ols).summary_frame(alpha=0.05)
    return pd.DataFrame({
        "Mes": base_2024["Mes"], "Fecha": base_2024["Fecha"] + pd.DateOffset(years=1),
        "Presupuesto": x["Presupuesto_marketing"],
        "Ventas_proyectadas": pred["mean"],
        "IC95_inf": pred["obs_ci_lower"], "IC95_sup": pred["obs_ci_upper"],   # intervalo de predicción mensual
        "Prob_meta": logistico_final.predict_proba(x)[:, 1],
        "Categoria": arbol_final.predict(x),
    }), X_ols

def intervalo_anual(X_ols, nivel=0.95):
    """Intervalo de predicción del TOTAL anual: suma de 12 meses nuevos (varianza de la suma)."""
    a = X_ols.sum().values                                   # suma de las filas de diseño
    var_total = a @ ols_final.cov_params().values @ a + len(X_ols) * ols_final.scale
    t = stats.t.ppf(0.5 + nivel / 2, ols_final.df_resid)
    total = float(a @ ols_final.params.values)
    return total - t * np.sqrt(var_total), total + t * np.sqrt(var_total)

proyecciones, filas = {}, []
for nombre, factor in ESCENARIOS.items():
    tabla, X_ols = proyectar(factor)
    proyecciones[nombre] = tabla
    inf, sup = intervalo_anual(X_ols)
    filas.append({"Escenario": nombre,
                  "Presupuesto anual": tabla["Presupuesto"].sum(),
                  "Ventas proyectadas": tabla["Ventas_proyectadas"].sum(),
                  "IC 95% inferior": inf, "IC 95% superior": sup,
                  "Meses con meta probable (>50 %)": int((tabla["Prob_meta"] > 0.5).sum()),
                  "Meses bajas / medias / altas": " / ".join(str((tabla["Categoria"] == c).sum())
                                                            for c in ["Bajas", "Medias", "Altas"])})

resumen_2025 = pd.DataFrame(filas).set_index("Escenario")
ventas_2024 = base_2024["Ventas_mensuales"].sum()
resumen_2025["Variación vs. 2024 (%)"] = (resumen_2025["Ventas proyectadas"] / ventas_2024 - 1) * 100
print(f"Ventas reales 2024: {ventas_2024:,.0f}")
resumen_2025
```

```python
fig, ax = plt.subplots(figsize=(12, 4.6))
historico = ventas[ventas["Año"] >= 2023]
ax.plot(historico["Fecha"], historico["Ventas_mensuales"], color="black", lw=1.8, label="Ventas reales 2023 – 2024")

colores_esc = {"Conservador": NARANJA, "Base": AZUL, "Expansión": VERDE_AGUA}
for nombre, tabla in proyecciones.items():
    ax.plot(tabla["Fecha"], tabla["Ventas_proyectadas"], color=colores_esc[nombre], lw=2,
            ls="--", marker="o", ms=4, label=f"Proyección 2025 – {nombre}")
base = proyecciones["Base"]
ax.fill_between(base["Fecha"], base["IC95_inf"], base["IC95_sup"], color=AZUL, alpha=0.12,
                label="Intervalo de predicción 95 % (escenario base)")
ax.axhline(UMBRAL_META, color=GRIS, ls=":", lw=1.2, label=f"Meta de ventas ({UMBRAL_META:,.0f})")
ax.axvline(pd.Timestamp("2025-01-01"), color=GRIS, lw=1)
ax.set_title("Proyección de ventas mensuales 2025 por escenario de inversión")
ax.set_ylabel("Ventas mensuales")
formato_miles(ax.yaxis)
ax.legend(frameon=False, fontsize=9, ncol=2, loc="upper left")
ax.set_ylim(top=ax.get_ylim()[1] * 1.18)
plt.tight_layout()
plt.show()
```

| P2 – Factores de la meta | Solo el presupuesto es determinante; sin él, promociones, nuevos clientes y opinión predicen casi al azar | Revisar la efectividad de las promociones antes de seguir invirtiendo en ellas |
| P3 – Categoría de ventas | El árbol acierta ≈ 89 % de los meses con reglas basadas en el presupuesto | Usar las reglas para preparar inventario y personal según el escenario esperado |
| P4 – Elección del modelo | La lineal es la más precisa y simple; logística y árbol aportan lectura de riesgo y de escenarios | Adoptar la regresión lineal como modelo principal y los otros dos como apoyo |
| P5 – Proyección 2025 | Repetir el presupuesto 2024 mantiene las ventas (≈ 93,300); ±10 % de inversión mueve las ventas ≈ ±9 – 10 % | Definir el presupuesto 2025 según el margen bruto: la expansión solo es rentable con margen > ≈ 66 % |

## 15. Conclusiones

1. **El análisis predictivo cumplió su propósito.** A partir de datos históricos se construyeron tres modelos que responden *qué podría ocurrir* con las ventas de la empresa de calzado, siguiendo los pasos que propone Google Cloud (s.f.): definir el problema, organizar y preprocesar los datos, desarrollar los modelos y validarlos.
2. **El presupuesto de marketing es el principal motor de las ventas.** Los tres modelos, construidos con técnicas diferentes, llegan a la misma conclusión: la correlación es de ≈ 0.98, la pendiente es de ≈ 1.58 y la variable concentra la mayor parte de la importancia en el árbol y en la regresión logística. Que tres enfoques coincidan refuerza la confianza en el hallazgo.
3. **Las promociones, los nuevos clientes y la opinión no mostraron un efecto significativo** en este histórico. No significa que no importen, sino que su efecto no se distingue del de la inversión en marketing. La opinión promedio baja (≈ 2.8 / 5) es además una alerta de servicio que conviene atender por sí misma.
4. **La validación fue clave para elegir y confiar en los modelos.** Separar datos de prueba y aplicar validación cruzada permitió detectar el sobreajuste del árbol con más de 3 niveles y comprobar que la regresión lineal y la logística son estables.
5. **La preparación de los datos sigue siendo indispensable.** Sin la corrección de los años 2068 – 2079 la serie temporal habría sido inconsistente. Como en el análisis descriptivo, la calidad del modelo depende de la calidad de los datos.
6. **El modelo se convierte en una herramienta de decisión.** La proyección 2025 traduce los resultados en escenarios concretos: con el mismo presupuesto de 2024 las ventas se mantendrían, y cada 10 % de inversión adicional las movería en una proporción similar. La decisión final depende del margen de la empresa, lo que conecta este análisis con el prescriptivo.
7. **Limitaciones.** El dataset es sintético y tiene solo 60 registros, por lo que las métricas tienen cierta variabilidad (lo evidencia la diferencia entre prueba y validación cruzada en el árbol). Los modelos muestran **asociaciones**, no causalidad, y no deben usarse con presupuestos fuera del rango observado. Para avanzar hacia el análisis **prescriptivo** (Universidad de Cundinamarca, 2024) sería útil incorporar más variables (precio, competencia, temporada) y nuevos periodos.
